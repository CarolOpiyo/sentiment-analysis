### Handling Duplicate Values

```python
# Count the number of duplicate rows
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")
```

```python
# Remove duplicate entries
df.drop_duplicates(inplace=True)
```

```python
# Defining stopwords set and initializing stemmer
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()

def clean_text(text):
    # Convert text to lowercase
    text = text.lower()
    # Remove URLs
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    # Remove user mentions and hashtags
    text = re.sub(r'\@\w+|\#', '', text)
    # Remove emoticons and other non-text characters
    text = re.sub(r'[\U00010000-\U0010ffff]', '', text, flags=re.UNICODE)  # For emoticons or other non-text characters
    # Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)
    # Remove numbers
    text = re.sub(r'\d+', '', text)
    # Tokenize text
    tokenized_text = word_tokenize(text)
    # Remove stop words and perform stemming
    filtered_text = [stemmer.stem(word) for word in tokenized_text if word not in stop_words]
    return " ".join(filtered_text)

# Apply the advanced cleaning function to your tweet text
df['cleaned_tweet_text'] = df['tweet_text'].apply(clean_text)
print(df[['tweet_text', 'cleaned_tweet_text']].head())
```

### Sentiment Analysis Plot

```python
# Initialize VADER Sentiment Intensity Analyzer
sid = SentimentIntensityAnalyzer()

# Apply sentiment analysis to each text
# This generates a dictionary of scores (neg, neu, pos, compound) for each text
df['sentiments'] = df['tweet_text'].apply(lambda x: sid.polarity_scores(x))

# Extract the compound score from the sentiments dictionary
df['compound'] = df['sentiments'].apply(lambda score_dict: score_dict['compound'])

# Initialize the sentiment_type column to empty strings
df['sentiment_type'] = ''

# Categorize sentiment based on the compound score
df.loc[df.compound > 0, 'sentiment_type'] = 'POSITIVE'  # Set sentiment_type to 'POSITIVE' for compound scores > 0
df.loc[df.compound == 0, 'sentiment_type'] = 'NEUTRAL'  # Set sentiment_type to 'NEUTRAL' for compound scores == 0
df.loc[df.compound < 0, 'sentiment_type'] = 'NEGATIVE'  # Set sentiment_type to 'NEGATIVE' for compound scores < 0

# Visualize the distribution of sentiment types
plt.figure(figsize=(10, 6))
sns.countplot(x='sentiment_type', data=df, order=['POSITIVE', 'NEUTRAL', 'NEGATIVE'])
plt.title('Sentiment Analysis Plot')
plt.show()
```

## Integrating BERT and ALBERT for sentiment analysis

```python
# labels for binary classification
df['label'] = (df['cyberbullying_type'] != 'not_cyberbullying').astype(int)
train_df, test_df = train_test_split(df, test_size=0.1, random_state=42)
```

### Tokenize Text Data

```python
class TweetDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len=128):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
    
    def __len__(self):
        return len(self.texts)
    
    def __getitem__(self, idx):
        text = self.texts[idx]
        label = self.labels[idx]
        encoding = self.tokenizer.encode_plus(
            text,
            add_special_tokens=True,
            max_length=self.max_len,
            return_token_type_ids=False,
            padding='max_length',
            return_attention_mask=True,
            return_tensors='pt',
            truncation=True
        )
        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'labels': torch.tensor(label, dtype=torch.long)
        }

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
train_dataset = TweetDataset(train_df['tweet_text'].tolist(), train_df['label'].tolist(), tokenizer)
test_dataset = TweetDataset(test_df['tweet_text'].tolist(), test_df['label'].tolist(), tokenizer)
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=16)
```

### Train BERT Model

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = BertForSequenceClassification.from_pretrained('bert-base-uncased', num_labels=2)
model.to(device)
optimizer = AdamW(model.parameters(), lr=2e-5)

epochs = 4  # Number of training epochs
model.train()

for epoch in range(epochs):
    for batch in train_loader:
        batch = {k: v.to(device) for k, v in batch.items()}
        outputs = model(input_ids=batch['input_ids'], attention_mask=batch['attention_mask'], labels=batch['labels'])
        loss = outputs.loss
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
    print(f"Epoch {epoch+1}/{epochs}, Loss: {loss.item()}")
```

### Bert Model Evaluation

```python
model.eval()
total_acc = 0
for batch in test_loader:
    batch = {k: v.to(device) for k, v in batch.items()}
    with torch.no_grad():
        outputs = model(input_ids=batch['input_ids'], attention_mask=batch['attention_mask'])
    logits = outputs.logits
    predictions = torch.argmax(logits, dim=-1)
    accuracy = (predictions == batch['labels']).float().mean()
    total_acc += accuracy.item()

print(f"Test Accuracy: {total_acc / len(test_loader)}")
```

## Model Deployment

We will use an Interactive Model Deployment with IPython Widgets and Joblib.

This deployment strategy employs Joblib and IPython Widgets to efficiently operationalize a machine learning model within a Jupyter Notebook. Joblib is chosen for its proficiency in handling large numpy arrays efficiently, making it ideal for serializing and reloading the logistic regression pipeline. This ensures quick and reliable model storage and access. Meanwhile, IPython Widgets provide an interactive user interface directly in the notebook. This allows for real-time input and output, enabling immediate testing and feedback. This combination of Joblib and IPython Widgets facilitates seamless integration and rapid prototyping within the familiar Jupyter environment, making it an excellent choice for demonstrations and educational purposes.

```python
# Save the pipeline to a file
joblib.dump(logistic_pipeline, 'logistic_pipeline.pkl')

# Load the pipeline back from the file
logistic_pipeline = joblib.load('logistic_pipeline.pkl')

# Use loaded file to make predictions
def predict_tweet(text):
    prediction = logistic_pipeline.predict([text])
    return prediction[0]

text = widgets.Textarea(
    value='Enter text here...',
    placeholder='Type something',
    description='Text:',
    disabled=False
)

button = widgets.Button(description="Predict")

output = widgets.Output()

def on_button_clicked(b):
    with output:
        output.clear_output()
        prediction = predict_tweet(text.value)
        print("Prediction:", prediction)

button.on_click(on_button_clicked)

display(text, button, output)
```

This is just a simple interface where users can type in a tweet, and it will display whether the tweet is considered cyberbullying and its specific type.
